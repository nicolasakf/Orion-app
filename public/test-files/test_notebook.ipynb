# Test Notebook

This notebook was created as a simple test notebook with a small pandas example and plot.

```
import pandas as pd
import matplotlib.pyplot as plt

print('Test notebook is ready!')
```

```
df = pd.DataFrame({
    'category': ['A', 'B', 'C', 'D'],
    'value': [10, 15, 7, 12]
})

df
```

```
ax = df.plot(kind='bar', x='category', y='value', legend=False, figsize=(6, 4))
ax.set_title('Sample Values by Category')
ax.set_xlabel('Category')
ax.set_ylabel('Value')
plt.tight_layout()
plt.show()
```
