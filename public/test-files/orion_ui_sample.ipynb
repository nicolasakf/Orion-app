# Orion UI Component Sample

This notebook demonstrates every current `orion_ui` primitive in one place.

- **Cell 1** builds a comprehensive UI showcase covering layouts, controls, display primitives, overlays, and advanced containers.
- **Cell 2** lists the primitives covered by the sample.
- **Cell 3** repeats the showcase output used by Orion App View.
- **Cell 4** retrieves the Python-side runtime state resulting from those controls.

*Note: In Orion App View or Notebook View, the UI cells render as rich native Orion components. Other frontends may show a static fallback.*

```
import orion_ui as ui

# App View points at this cell's output, so keep the comprehensive showcase here.
# The code mirrors Cell 1 without autoreload setup.

text_input = ui.input("sample_text", label="Text Input", placeholder="Type something...", default_value="Hello Orion")
email_input = ui.input("sample_email", label="Email Input", input_type="email", placeholder="name@example.com", default_value="hello@orion.dev")
textarea_input = ui.textarea("sample_textarea", label="Text Area", placeholder="Multi-line text...", default_value="This is a textarea.")
select_input = ui.select(
    "sample_select",
    [{"label": "Option A", "value": "option_a"}, {"label": "Option B", "value": "option_b"}, {"label": "Option C", "value": "option_c"}],
    label="Select Box",
    default_value="option_a",
    placeholder="Choose an option",
)
slider_input = ui.slider("sample_slider", label="Range Slider", min=0, max=100, default_value=75, step=5)
switch_input = ui.switch("sample_switch", label="Toggle Switch", default_value=True)
checkbox_input = ui.checkbox("sample_checkbox", label="Checkbox", default_value=True)
radio_input = ui.radio_group("sample_radio", ["Small", "Medium", "Large"], label="Radio Group", default_value="Medium")
toggle_input = ui.toggle("sample_toggle", label="Single Toggle", default_value=True, variant="outline")
toggle_group_input = ui.toggle_group(
    "sample_toggle_group",
    [{"label": "Daily", "value": "daily"}, {"label": "Weekly", "value": "weekly"}, {"label": "Monthly", "value": "monthly"}],
    label="Toggle Group",
    default_value="weekly",
    variant="outline",
)
calendar_input = ui.calendar("sample_calendar", label="Calendar", default_value="2026-05-27")
date_picker_input = ui.date_picker("sample_date_picker", label="Date Picker", default_value="2026-06-01", placeholder="Pick a date")
progress_bound = ui.progress("sample_progress", label="Bound Progress", default_value=60, max=100)

controls_card = ui.card(
    ui.stack(
        ui.grid(text_input, email_input, columns=2),
        textarea_input,
        ui.grid(select_input, slider_input, columns=2),
        ui.grid(switch_input, checkbox_input, columns=2),
        ui.grid(radio_input, toggle_input, toggle_group_input, columns=3),
        ui.grid(calendar_input, date_picker_input, columns=2),
        progress_bound,
        gap="md",
    ),
    title="Controls",
    description="Inputs and state-bound controls, including radio, toggle, calendar, date picker, and progress primitives.",
)

display_card = ui.card(
    ui.stack(
        ui.label("Labels render plain text inside Orion UI layouts."),
        ui.grid(ui.badge("Default"), ui.badge("Secondary", variant="secondary"), ui.badge("Destructive", variant="destructive"), columns=3, gap="sm"),
        ui.grid(ui.button("Default Button"), ui.button("Secondary", variant="secondary", size="sm"), ui.button("Destructive", variant="destructive"), columns=3, gap="sm"),
        ui.alert(title="Alert", description="Alerts highlight important status or guidance."),
        ui.alert(title="Destructive Alert", description="Use variants for stronger visual emphasis.", variant="destructive"),
        ui.progress(label="Static Progress", default_value=35, max=100),
        ui.grid(
            ui.avatar(src="https://github.com/github.png", alt="GitHub logo", fallback="GH", label="Image avatar", size="md"),
            columns=2,
            gap="sm",
        ),
        ui.separator(),
        ui.label("End of display primitives."),
        gap="md",
    ),
    title="Displays & Actions",
    description="Labels, badges, buttons, alerts, progress bars, avatars, and separators.",
)

overlays_card = ui.card(
    ui.stack(
        ui.grid(
            ui.popover(ui.label("Popover child content can contain any Orion UI component."), label="Open Popover", description="A click-triggered floating panel."),
            ui.hover_card(ui.label("Hover cards appear when the trigger is hovered."), label="Hover Card", description="Useful for contextual previews."),
            ui.tooltip(ui.label("Tooltip details live here."), label="Tooltip", description="Short hover or focus help text."),
            columns=3,
            gap="sm",
        ),
        ui.collapsible(ui.label("This content starts open and can be collapsed."), title="Collapsible Primitive", description="Disclosure container with optional child content.", default_open=True),
        gap="md",
    ),
    title="Overlays & Disclosure",
    description="Popover, hover card, tooltip, and collapsible primitives.",
)

tabs_demo = ui.tabs(
    ui.card(ui.label("First tab panel content."), title="Overview"),
    ui.card(ui.label("Second tab panel content."), title="Details"),
    ui.card(ui.label("Third tab panel content."), title="Settings"),
    default_value="Overview",
)
accordion_demo = ui.accordion(
    ui.card(ui.label("Accordion item one content."), title="Accordion Item 1"),
    ui.card(ui.label("Accordion item two content."), title="Accordion Item 2"),
    ui.card(ui.label("Accordion item three content."), title="Accordion Item 3"),
    multiple=True,
)
carousel_demo = ui.carousel(
    ui.card(ui.label("Slide 1"), description="Carousel slide content."),
    ui.card(ui.label("Slide 2"), description="More slide content."),
    ui.card(ui.label("Slide 3"), description="Final slide content."),
    show_controls=True,
)
containers_card = ui.card(
    ui.stack(ui.section(tabs_demo, title="Tabs"), ui.section(accordion_demo, title="Accordion"), ui.section(carousel_demo, title="Carousel"), gap="lg"),
    title="Advanced Containers",
    description="Tabs, accordion, carousel, sections, cards, stacks, and grids.",
)

notebook_card = ui.card(
    ui.stack(
        ui.markdown_cell(text="### Inline MarkdownCell\nThis MarkdownCell is rendered from inline markdown text."),
        ui.markdown_cell(cell_id="fa981ee1-a64d-4b22-bb99-f3b1a541814a"),
        ui.alert(title="Output primitive", description="App View itself uses an Output primitive to display this cell's Orion UI result."),
        gap="md",
    ),
    title="Notebook-aware Primitives",
    description="MarkdownCell examples plus a note about Output primitives.",
)

demo_page = ui.page(
    ui.section(
        ui.grid(controls_card, display_card, columns=1, gap="lg"),
        ui.grid(overlays_card, containers_card, notebook_card, columns=1, gap="lg"),
        title="Orion UI Component Showcase",
        description="A comprehensive page featuring all currently available Orion UI primitives.",
        padding="md",
        gap="lg",
    ),
    gap="lg",
    padding="md",
)

demo_page
```

```
import orion_ui as ui

# To retrieve the current values from the UI controls, use `ui.get()` or `ui.state()`.
# When the user interacts with the UI above, the Python-side state is automatically updated.

state_keys = [
    "sample_text",
    "sample_email",
    "sample_textarea",
    "sample_select",
    "sample_slider",
    "sample_switch",
    "sample_checkbox",
    "sample_radio",
    "sample_toggle",
    "sample_toggle_group",
    "sample_calendar",
    "sample_date_picker",
    "sample_progress",
]

print("Individual values:")
for key in state_keys:
    print(f"{key}: {ui.get(key)}")

print("\nFull UI State dictionary:")
print(ui.state())
```
