# DataFrame shapes for table extractor verification

This notebook creates various pandas DataFrame configurations to verify `extractTableFromHTML` correctly parses pandas `_repr_html_` output.

Run each cell and compare the displayed table with Orion's rendered output.

```python
import pandas as pd
```

## 1. Simple single-index (default integer index, no names)

```python
df1 = pd.DataFrame({"A": [1, 2], "B": [3, 4]})
df1
```

## 2. Named index, unnamed columns

```python
df2 = pd.DataFrame({"A": [1, 2], "B": [3, 4]}, index=["x", "y"])
df2.index.name = "my_index"
df2
```

## 3. Unnamed index, named columns

```python
df3 = pd.DataFrame({"A": [1, 2], "B": [3, 4]})
df3.columns.name = "my_cols"
df3
```

## 4. Named index AND named columns (bug case — 2-row thead)

```python
df4 = pd.DataFrame({"A": [1, 2], "B": [3, 4]}, index=["x", "y"])
df4.index.name = "my_index"
df4.columns.name = "my_cols"
df4
```

## 5. Multi-index columns, single row index

```python
df5 = pd.DataFrame(
    [[1, 2, 3, 4], [5, 6, 7, 8]],
    columns=pd.MultiIndex.from_tuples([("A", "one"), ("A", "two"), ("B", "one"), ("B", "two")]),
)
df5
```

## 6. Multi-index rows (unnamed levels)

```python
df6 = pd.DataFrame(
    [[1, 2, 3], [4, 5, 6], [7, 8, 9]],
    index=pd.MultiIndex.from_tuples([("bar", "one"), ("bar", "two"), ("baz", "one")]),
    columns=pd.MultiIndex.from_tuples([("A", "x"), ("A", "y"), ("B", "x")]),
)
df6
```

## 7. Named multi-index rows

```python
df7 = pd.DataFrame(
    [[100, 200], [110, 210], [300, 400]],
    index=pd.MultiIndex.from_tuples(
        [("US", "2020"), ("US", "2021"), ("EU", "2020")],
        names=["Region", "Year"],
    ),
    columns=["val1", "val2"],
)
df7
```

## 8. No index (index=False)

```python
df8 = pd.DataFrame({"A": [1, 4], "B": [2, 5], "C": [3, 6]})
df8
```

```python
# Display without index (use to_html for raw HTML)
from IPython.display import HTML
HTML(df8.to_html(index=False))
```

## 9. Three-level column MultiIndex

```python
df9 = pd.DataFrame(
    [[1, 2, 3, 4]],
    columns=pd.MultiIndex.from_tuples(
        [("A", "x", "i"), ("A", "x", "ii"), ("A", "y", "i"), ("B", "x", "i")]
    ),
)
df9
```

## 10. Raw HTML snippets for extractTableFromHTML

```python
# Print _repr_html_ for each DataFrame (for copy-paste into tests)
for name, html in [
    ("df1", df1._repr_html_()),
    ("df2", df2._repr_html_()),
    ("df3", df3._repr_html_()),
    ("df4", df4._repr_html_()),
    ("df5", df5._repr_html_()),
    ("df6", df6._repr_html_()),
    ("df7", df7._repr_html_()),
    ("df8 (no index)", df8.to_html(index=False)),
    ("df9", df9._repr_html_()),
]:
    print(f"\n=== {name} ===")
    print(html)
```
