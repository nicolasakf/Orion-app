# Text Outputs Demo
This notebook demonstrates various types, sizes, and colors of text outputs.

```python
import sys

# Standard output (stdout)
print("This is a standard print statement (stdout).")

# Standard error (stderr)
sys.stderr.write("This is a standard error statement (stderr).\n")
```

```python
# ANSI Escape Codes for terminal colors and styles
print("\033[31mThis is red text\033[0m")
print("\033[32mThis is green text\033[0m")
print("\033[33mThis is yellow text\033[0m")
print("\033[34mThis is blue text\033[0m")
print("\033[35mThis is magenta text\033[0m")
print("\033[36mThis is cyan text\033[0m")
print("\033[1mThis is bold text\033[0m")
print("\033[4mThis is underlined text\033[0m")
print("\033[43m\033[30mThis has a yellow background and black text\033[0m")
```

```python
from IPython.display import display, HTML, Markdown

# Sizes and semantic headings using HTML
display(HTML("<h1>Heading 1 (Largest)</h1>"))
display(HTML("<h2>Heading 2</h2>"))
display(HTML("<h3>Heading 3</h3>"))
display(HTML("<h4>Heading 4</h4>"))
display(HTML("<h5>Heading 5</h5>"))
display(HTML("<h6>Heading 6 (Smallest)</h6>"))

# Specific font sizes
display(HTML("<p style='font-size: 10px;'>This is 10px text (Small).</p>"))
display(HTML("<p style='font-size: 16px;'>This is 16px text (Normal).</p>"))
display(HTML("<p style='font-size: 24px;'>This is 24px text (Large).</p>"))
```

```python
# Very long text output
for i in range(1, 1001):
    print(f"This is line number {i} in a very long output stream to demonstrate scrolling or truncation in the notebook interface.")
```

```python
# Colors and styles using HTML and Markdown
display(HTML("<p style='color: magenta;'>This is magenta text using HTML.</p>"))
display(HTML("<p style='color: cyan; background-color: black; padding: 5px;'>Cyan text on a black background</p>"))

# Markdown styling
display(Markdown("**This is bold Markdown text**"))
display(Markdown("*This is italic Markdown text*"))
display(Markdown("`This is inline code text`"))
display(Markdown("<span style='color:orange; font-size:20px'>This is orange 20px text via inline HTML in Markdown</span>"))
```
